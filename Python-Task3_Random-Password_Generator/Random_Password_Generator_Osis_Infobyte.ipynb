```python
# Random Password Generator
# Oasis Infobyte - Python Programming Task 3

import string
import random

print("===== RANDOM PASSWORD GENERATOR =====")

try:
    # Password length
    length = int(input("Enter password length (minimum 8): "))

    if length < 8:
        print("Error: Password length must be at least 8.")

    else:
        print("\nSelect at least 2 character types:")
        print("1. Letters")
        print("2. Numbers")
        print("3. Special characters")

        choice = input("Enter your choices (e.g., 123): ")

        # Remove duplicate choices
        choice = set(choice)

        # Validate choices
        if len(choice) < 2 or not choice.issubset({"1", "2", "3"}):
            print("Error: Please select at least 2 valid character types.")

        else:
            characters = ""

            if "1" in choice:
                characters += string.ascii_letters

            if "2" in choice:
                characters += string.digits

            if "3" in choice:
                characters += string.punctuation

            # Generate password
            password = ''.join(
                random.choice(characters) for _ in range(length)
            )

            print("\nGenerated Password:", password)

            # Regenerate password
            again = input(
                "\nDo you want to generate another password? (yes/no): "
            )

            if again.lower() == "yes":
                new_password = ''.join(
                    random.choice(characters) for _ in range(length)
                )
                print("New Password:", new_password)

            else:
                print("Thank you for using the Password Generator!")

except ValueError:
    print("Error: Please enter a valid number.")
```
